### Missing Value Treatment

Missing values were analyzed based on the meaning of each variable.

Therefore, these null values were **not removed or replaced**, as they represent valid and meaningful missing values in the cricket dataset.

```python
df[df['winner'].isnull()][
    ['match_id', 'inning', 'batting_team', 'result']
].drop_duplicates().sort_values(['match_id', 'inning'])
```

```python
last_inning_team = (
    df.groupby('match_id')
      .apply(lambda x: x.loc[x['inning'].idxmax(), 'batting_team'])
)

known_winner = (
    df[['match_id', 'winner']]
      .dropna(subset=['winner'])
      .drop_duplicates('match_id')
)

check = known_winner.copy()

check['last_inning_team'] = check['match_id'].map(last_inning_team)

check['matches'] = check['winner'] == check['last_inning_team']

check['matches'].value_counts()
```

```python
match_info = df[['match_id', 'winner', 'result']].drop_duplicates('match_id')

second_inning_team = (
    df[df['inning'] == 2]
    .groupby('match_id')['batting_team']
    .first()
)

check = match_info[
    (match_info['result'] == 'wickets') &
    (match_info['winner'].notna())
].copy()

check['second_inning_team'] = check['match_id'].map(second_inning_team)

check['winner_matches'] = (
    check['winner'] == check['second_inning_team']
)

check['winner_matches'].value_counts()
```

```python
second_inning_team = (
    df[df['inning'] == 2]
    .groupby('match_id')['batting_team']
    .first()
)

df['winner'] = df['winner'].fillna(
    df['match_id'].map(second_inning_team)
)
```

```python
df['winner'].isnull().sum()
```

```python
remaining_winner = df[df['winner'].isnull()][
    ['match_id', 'inning', 'batting_team', 'team1', 'team2', 'result']
].drop_duplicates().sort_values(['match_id', 'inning'])

remaining_winner
```

```python
df[df['winner'].isnull()]['inning'].value_counts()
```

```python
# 1. Check remaining missing values
print("Missing winner values:", df['winner'].isnull().sum())

# 2. Check that each match has at most one unique winner
winner_per_match = df.groupby('match_id')['winner'].nunique(dropna=True)
print("Maximum unique winners in a match:", winner_per_match.max())

# 3. Check whether winner is one of the two teams
valid_winner = (
    df['winner'].isna() |
    (df['winner'] == df['team1']) |
    (df['winner'] == df['team2'])
)

print("Invalid winner records:", (~valid_winner).sum())

# 4. Check filled winners against second innings
second_inning_team = (
    df[df['inning'] == 2]
    .groupby('match_id')['batting_team']
    .first()
)

wicket_matches = df[
    (df['result'] == 'wickets') &
    (df['winner'].notna())
].drop_duplicates('match_id').copy()

wicket_matches['second_inning_team'] = (
    wicket_matches['match_id'].map(second_inning_team)
)

wicket_matches['verified'] = (
    wicket_matches['winner'] == wicket_matches['second_inning_team']
)

print("Wicket-result matches verified:")
print(wicket_matches['verified'].value_counts(dropna=False))
```

### Winner Column – Missing Value Treatment

The **`winner`** column was analyzed using **`match_id`**, **`inning`**, and **`batting_team`**.

Missing winner values were filled where the second-innings batting team could reliably determine the match winner. This resulted in **4,207 values being filled**.

A total of **495 values** remained missing because only first-innings information was available, making it impossible to determine the winner reliably. These values were retained as `NaN` rather than making assumptions.

The cleaned column was verified to ensure that each match has at most one winner and that all available winners belong to either **`team1`** or **`team2`**.
