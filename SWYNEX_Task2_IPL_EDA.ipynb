```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('ipl_cleaned_new.csv')
pd.set_option('display.max_columns', None)
df.head()
```

```python
print("Number of Rows:",df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
missing_values = df.isnull().sum()

missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
df['match_id'] = df['match_id'].astype('category')
```

```python
df.dtypes
```

```python
categorical_cols = [
    'season', 'city', 'venue', 'team1', 'team2',
    'toss_winner', 'toss_decision', 'winner',
    'result', 'player_of_match', 'batting_team',
    'bowling_team', 'batter', 'bowler',
    'non_striker', 'wicket_kind', 'player_out', 'fielder'
]

df[categorical_cols].dtypes
```

```python
## Season match count
season_matches = df.groupby('season')['match_id'].nunique()

season_matches
```

```python
plt.figure(figsize=(12, 5))

sns.barplot(x=season_matches.index, y=season_matches.values)

plt.title('Number of Matches Played per Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')

plt.xticks(rotation=45)
plt.show()
```

```python
## Seson wise total runs
season_runs = df.groupby('season')['runs_scored'].sum()

season_runs
```

```python
plt.figure(figsize=(12, 5))

sns.barplot(x=season_runs.index, y=season_runs.values)

plt.title('Total Runs Scored per Season')
plt.xlabel('Season')
plt.ylabel('Total Runs')

plt.xticks(rotation=45)
plt.show()
```

Total runs vary across IPL seasons, with the highest aggregate runs observed in 2025, followed by 2023 and 2024. Seasons with fewer matches, such as 2018 and 2021, also show lower total runs.

```python
team_sixes = (
    df[df['runs_scored'] == 6]
    .groupby('batting_team')
    .size()
    .sort_values(ascending=False)
)

team_sixes
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=team_sixes.values,
    y=team_sixes.index,
    palette='husl',
    hue=team_sixes.index,
    legend=False
)

plt.title('Total Sixes Hit by Each Team')
plt.xlabel('Number of Sixes')
plt.ylabel('Team')

plt.show()
```

```python
## Top 10 scorers 
top_10_run_scorers = (
    df.groupby('batter')['runs_scored']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_10_run_scorers
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_run_scorers.values,
    y=top_10_run_scorers.index,
    palette='husl',
    hue=top_10_run_scorers.index,
    legend=False
)

plt.title('Top 10 Players by Total Runs')
plt.xlabel('Total Runs')
plt.ylabel('Player')

plt.show()
```

```python
## top 10 wicket takers
bowler_credited_wickets = [
    'bowled',
    'caught',
    'caught and bowled',
    'lbw',
    'stumped',
    'hit wicket'
]

top_10_wicket_takers = (
    df[df['wicket_kind'].isin(bowler_credited_wickets)]
    .groupby('bowler')
    .size()
    .sort_values(ascending=False)
    .head(10)
)

top_10_wicket_takers
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_wicket_takers.values,
    y=top_10_wicket_takers.index,
    palette='husl',
    hue=top_10_wicket_takers.index,
    legend=False
)

plt.title('Top 10 Wicket Takers')
plt.xlabel('Total Wickets')
plt.ylabel('Bowler')

plt.show()
```

```python
## numerical correltion
numeric_cols = df.select_dtypes(include=np.number)

correlation = numeric_cols.corr()

correlation
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

Insight: The correlation analysis shows strong relationships among variables that represent innings progression, particularly overs, current score, balls remaining, and wickets remaining. In contrast, individual runs scored show relatively weak correlations with most match-state variables.

```python
## Current score vs current run rate
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='current_run_rate',
    y='current_score',
    alpha=0.3
)

plt.title('Current Score vs Current Run Rate')
plt.xlabel('Current Run Rate')
plt.ylabel('Current Score')

plt.show()
```

```python
## Outliers analysis of Runs scored
plt.figure(figsize=(10, 5))

sns.boxplot(x=df['runs_scored'])

plt.title('Boxplot of Runs Scored per Ball')
plt.xlabel('Runs Scored')

plt.show()
```

Insight: Most deliveries result in 0–1 runs, while higher scoring deliveries such as 3, 4, 5 and 6 runs appear as statistical outliers. These are valid cricket scoring outcomes rather than data-quality issues.

## Final EDA Insights & Conclusion

### Key Insights

1. The number of matches varies across IPL seasons, with some seasons having significantly more matches than others.

2. Total runs also vary across seasons. Seasons with a higher number of matches generally show higher aggregate runs.

3. The analysis of team-wise sixes shows considerable variation in six-hitting across teams.

4. A small group of players contributes a large share of the total runs, while the top wicket-takers account for a significant number of bowler-credited dismissals.

5. The correlation analysis shows strong relationships between innings progression variables such as overs, current score, balls remaining, and wickets remaining.

6. Current score and current run rate do not show a strong linear relationship, indicating that similar run rates can occur at different stages of an innings.

7. Most deliveries result in 0 or 1 run. Values such as 3, 4, 5, and 6 appear as statistical outliers in the boxplot, but these are valid cricket scoring outcomes rather than data errors.

### Conclusion

The Exploratory Data Analysis provided an overall understanding of IPL match, team, player, scoring, and innings-related patterns. The analysis also identified relationships between important numerical variables and highlighted the distribution of runs scored per delivery. The cleaned dataset is therefore suitable for further analysis and visualization.
